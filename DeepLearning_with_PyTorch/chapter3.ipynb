```python
img_t = torch.randn(3, 5, 5)
print(img_t)
weights = torch.tensor([0.2126, 0.7152, 0.0722])
print(weights)
```

```python
batch_t = torch.randn(2, 3, 5, 5)
print(batch_t)
```

```python
img_gray_naive = img_t.mean(-3)
print(img_gray_naive)
batch_gray_naive = batch_t.mean(-3)
print(batch_gray_naive)
print(img_gray_naive.shape, batch_gray_naive.shape)
```

```python
import matplotlib.pyplot as plt

# Assuming img_gray_naive is a 2D tensor (height x width)
plt.imshow(img_gray_naive, cmap='gray')
plt.title('Grayscale Image (Naive Mean)')
plt.colorbar()
plt.show()
```

```python
unsqueezed_weights = weights.unsqueeze(-1).unsqueeze_(-1)
print(unsqueezed_weights)
```

```python
img_weights = (img_t * unsqueezed_weights)
print(img_weights)
batch_weights = (batch_t * unsqueezed_weights)
print(batch_weights)
```

```python
img_gray_weighted = img_weights.sum(-3)
print(img_gray_weighted)
batch_gray_weighted = batch_weights.sum(-3)
print(batch_gray_weighted)
print(img_gray_weighted.shape,
      batch_gray_weighted.shape,
      batch_weights.shape,
      batch_t.shape, unsqueezed_weights.shape)
```

```python
img_gray_weighted_fancy = torch.einsum('...chw,c->...hw', img_t, weights)
batch_gray_weighted_fancy = torch.einsum('...chw,c->...hw', batch_t, weights)
print(batch_gray_weighted_fancy.shape)
```

```python
weights_named = torch.tensor([0.2126, 0.7152, 0.0722], names=['channels'])
weights_named
```

```python
img_named = img_t.refine_names(..., 'channels', 'rows', 'columns')
batch_named = batch_t.refine_names(..., 'channels', 'rows', 'columns')
print("img named:", img_named.shape, img_named.names)
print("batch named:", batch_named.shape, batch_named.names)
```

```python
weights_aligned = weights_named.align_as(img_named)
weights_aligned.shape, weights_aligned.names
```

```python
gray_named = (img_named * weights_aligned).sum('channels')
gray_named.shape, gray_named.names
```

```python
gray_named = (img_named[..., :3] * weights_named).sum('channels')
```

```python
gray_plain = gray_named.rename(None)
gray_plain.shape, gray_plain.names
```
