```python
#Addition, subtraction, multiplication, exponentiation

print("x + 3 is", x + 3)
print("x - 15 is", x - 15)
print("x * 3 is", x * 3)
print("x / 2 is", x / 2)
print("x // 2 (floor division) is", x // 2)
print("x mod 2 (remainder) is", x % 2)
print("x ** 3 (exponentiation) is", x ** 3)

x += 1
print("x += 1 is", x)
x *= 2
print("x *= 2 is", x)
```
